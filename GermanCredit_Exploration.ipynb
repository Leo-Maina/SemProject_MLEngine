## 1. Open-Source Dataset Discovery
To develop our credit risk evaluation engine, we identified three open-source datasets in our research area:

a) **German Credit Risk Dataset (UCI Machine Learning Repository)**
   * **Description**: Contains 1,000 credit records with demographic, financial, and transactional attributes.
   * **Target Variable**: Column `24` / `credit_risk` (1 = Good Credit Risk, 2 = Bad Credit Risk)

b) **Kaggle Loan Approval Prediction Dataset**
   * **Description**: Features applicant income, loan amount, credit history, and demographic data.
   * **Target Variable**: `Loan_Status` (Y = Approved, N = Rejected)

c) **Give Me Some Credit Dataset (Kaggle)**
   * **Description**: Contains financial data for over 150,000 borrowers predicting financial distress.
   * **Target Variable**: `SeriousDlqin2yrs` (1 = 90+ days past due delinquency, 0 = Otherwise)


## 2. Selected Dataset Choice
We selected the **UCI German Credit Risk Dataset**. It provides complete numerical data representing financial and demographic attributes, making it ideal for evaluating micro-loan creditworthiness in a mobile application backend.

Cell 1: Setup & Imports

```python
import pandas as pd
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/statlog/german/german.data-numeric"
df = pd.read_csv(url, sep=r'\s+', header=None)
df.head()
```

Cell 2: Question (a) - Row & Column Count

```python
num_rows, num_cols = df.shape
print(f"Dataset Shape: {num_rows} rows and {num_cols} columns")
print(f"Total Rows: {num_rows}")
print(f"Total Columns: {num_cols}")
```

Cell 3: Question (b) - Data Types

```python
print("Data Types contained in the dataset:")
print(df.dtypes)
print("\nSummary of unique data types:")
print(df.dtypes.value_counts())
```

Cell 4: Question (c) - Completeness / Missing Values

```python
missing_counts = df.isnull().sum()
total_missing = missing_counts.sum()
print("Missing values per column:")
print(missing_counts[missing_counts > 0])
print("\n--- Completeness Check ---")
if total_missing == 0:
    print("Yes, the dataset is complete. There are no missing values.")
else:
    print(f"No, the dataset is NOT complete. It contains {total_missing} missing values in total.")
```
